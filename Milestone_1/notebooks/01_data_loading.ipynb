## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')

print('Libraries loaded successfully!')
```

## Step 2: Define File Paths

```python
RAW_DATA_PATH = '../data/raw/'

# Dataset 1 - Hospital Management System (Core/HMIS)
HMIS_FILE = RAW_DATA_PATH + 'Hospital_Management_System.xlsx'

# Dataset 2 - Hospital Beds Management
BEDS_PATIENTS_FILE     = RAW_DATA_PATH + 'beds_patients.csv'
BEDS_SERVICES_FILE     = RAW_DATA_PATH + 'beds_services_weekly.csv'
BEDS_STAFF_FILE        = RAW_DATA_PATH + 'beds_staff.csv'
BEDS_SCHEDULE_FILE     = RAW_DATA_PATH + 'beds_staff_schedule.csv'

# Dataset 3 - Hospital Readmission
READMISSION_FILE       = RAW_DATA_PATH + 'readmission_admission_data.csv'
MORTALITY_FILE         = RAW_DATA_PATH + 'readmission_mortality_data.csv'

# Dataset 4 - Healthcare Dataset
HEALTHCARE_FILE        = RAW_DATA_PATH + 'healthcare_dataset.csv'

print('File paths defined!')
```

## Step 3: Load Dataset 1 – Hospital Management System (CORE)

```python
# Load all sheets from HMIS Excel file
hmis_sheets = pd.read_excel(HMIS_FILE, sheet_name=None)

print('HMIS Sheets Available:')
for sheet_name, df in hmis_sheets.items():
    print(f'  - {sheet_name}: {df.shape[0]} rows x {df.shape[1]} columns')
```

```python
# Extract individual tables from HMIS
df_appointments  = hmis_sheets['Appointment']
df_bed           = hmis_sheets['Bed']
df_bed_records   = hmis_sheets['BedRecords']
df_department    = hmis_sheets['Department']
df_doctor        = hmis_sheets['Doctor']
df_helpers       = hmis_sheets['Helpers']
df_medical       = hmis_sheets['MedicalRecord']
df_nurse         = hmis_sheets['Nurse']
df_patients      = hmis_sheets['Patients']
df_room          = hmis_sheets['Room']
df_room_records  = hmis_sheets['RoomRecords']
df_staff_shift   = hmis_sheets['StaffShift']
df_surgery       = hmis_sheets['SurgeryRecord']
df_ward          = hmis_sheets['Ward']

print('All HMIS tables loaded!')
```

```python
# Preview key tables
print('--- Patients Table ---')
display(df_patients.head(3))

print('--- Department Table ---')
display(df_department.head(3))

print('--- BedRecords Table (Admissions) ---')
display(df_bed_records.head(3))

print('--- Ward Table ---')
display(df_ward.head(3))
```

## Step 4: Load Dataset 2 – Hospital Beds Management

```python
df_beds_patients  = pd.read_csv(BEDS_PATIENTS_FILE)
df_beds_services  = pd.read_csv(BEDS_SERVICES_FILE)
df_beds_staff     = pd.read_csv(BEDS_STAFF_FILE)
df_beds_schedule  = pd.read_csv(BEDS_SCHEDULE_FILE)

print('Dataset 2 - Beds Management Loaded!')
print(f'  beds_patients:  {df_beds_patients.shape}')
print(f'  beds_services:  {df_beds_services.shape}')
print(f'  beds_staff:     {df_beds_staff.shape}')
print(f'  beds_schedule:  {df_beds_schedule.shape}')
```

```python
print('--- Beds Patients ---')
display(df_beds_patients.head(3))

print('--- Beds Services Weekly ---')
display(df_beds_services.head(3))
```

## Step 5: Load Dataset 3 – Hospital Readmission

```python
df_readmission = pd.read_csv(READMISSION_FILE)
df_mortality   = pd.read_csv(MORTALITY_FILE)

print('Dataset 3 - Readmission Loaded!')
print(f'  Admission Data: {df_readmission.shape}')
print(f'  Mortality Data: {df_mortality.shape}')
```

```python
print('--- Readmission Admission Data ---')
display(df_readmission.head(3))
print('Columns:', list(df_readmission.columns))
```

## Step 6: Load Dataset 4 – Healthcare Dataset

```python
df_healthcare = pd.read_csv(HEALTHCARE_FILE)

print('Dataset 4 - Healthcare Dataset Loaded!')
print(f'  Shape: {df_healthcare.shape}')
display(df_healthcare.head(3))
```

## Step 7: Summary of All Datasets

```python
summary = {
    'Dataset': [
        'HMIS - Appointments', 'HMIS - BedRecords', 'HMIS - Department',
        'HMIS - Doctor', 'HMIS - Nurse', 'HMIS - Patients',
        'HMIS - Ward', 'HMIS - Bed', 'HMIS - MedicalRecord',
        'HMIS - StaffShift', 'HMIS - SurgeryRecord',
        'Beds - Patients', 'Beds - Services Weekly',
        'Beds - Staff', 'Beds - Schedule',
        'Readmission - Admission', 'Readmission - Mortality',
        'Healthcare Dataset'
    ],
    'Rows': [
        len(df_appointments), len(df_bed_records), len(df_department),
        len(df_doctor), len(df_nurse), len(df_patients),
        len(df_ward), len(df_bed), len(df_medical),
        len(df_staff_shift), len(df_surgery),
        len(df_beds_patients), len(df_beds_services),
        len(df_beds_staff), len(df_beds_schedule),
        len(df_readmission), len(df_mortality),
        len(df_healthcare)
    ],
    'Columns': [
        df_appointments.shape[1], df_bed_records.shape[1], df_department.shape[1],
        df_doctor.shape[1], df_nurse.shape[1], df_patients.shape[1],
        df_ward.shape[1], df_bed.shape[1], df_medical.shape[1],
        df_staff_shift.shape[1], df_surgery.shape[1],
        df_beds_patients.shape[1], df_beds_services.shape[1],
        df_beds_staff.shape[1], df_beds_schedule.shape[1],
        df_readmission.shape[1], df_mortality.shape[1],
        df_healthcare.shape[1]
    ]
}

summary_df = pd.DataFrame(summary)
print('=== COMPLETE DATASET SUMMARY ===')
display(summary_df)
print(f'\nTotal Tables: {len(summary_df)}')
print(f'Total Records: {summary_df["Rows"].sum():,}')
```
